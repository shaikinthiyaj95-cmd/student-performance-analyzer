```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
 
sns.set_theme(style="whitegrid", context="notebook")
 
GRADE_BINS = [-np.inf, 50, 60, 70, 80, 90, np.inf]
GRADE_LABELS = ["F", "D", "C", "B", "A", "A+"]
GRADE_ORDER = ["A+", "A", "B", "C", "D", "F"]
RESULT_COLORS = {"PASS": "#2a9d8f", "FAIL": "#e63946"}
 
# Words that mark a numeric column as "not a subject"
NON_SUBJECT_TOKENS = {
    "id", "roll", "rollno", "no", "number", "age", "total", "sum", "avg",
    "average", "mean", "pct", "percent", "percentage", "rank", "attendance",
    "attend", "grade", "result", "gpa", "cgpa", "sno", "sl",
}
 
 
def tokens(col: str) -> set:
    cleaned = "".join(ch if ch.isalnum() else " " for ch in str(col).lower())
    return set(cleaned.split())
 
 
class StudentPerformanceAnalyzer:
    def __init__(self, path=None, pass_mark=50.0,
                 subject_pass=40.0, max_marks=100.0, att_threshold=75.0,
                 show=True):
        self.path = path
        self.pass_mark = pass_mark
        self.subject_pass = subject_pass
        self.max_marks = max_marks
        self.att_threshold = att_threshold
        self.show = show
 
        self.df = None
        self.name_col = None
        self.att_col = None
        self.subject_cols = []
        self.subject_stats = None
        self.warnings = []
        self.insights = {}
 
    # ------------------------------------------------------------------ #
    # 1. LOAD
    # ------------------------------------------------------------------ #
    def load_data(self):
        print("=" * 64)
        print("        STUDENT PERFORMANCE ANALYZER (ADVANCED)")
        print("=" * 64)
 
        if not self.path:
            self.path = input("Enter CSV file path (e.g. student_data.csv): ")
        self.path = str(self.path).strip().strip("'\"")
 
        df = None
        for enc in ("utf-8", "utf-8-sig", "latin-1"):
            try:
                df = pd.read_csv(self.path, encoding=enc, skipinitialspace=True)
                break
            except UnicodeDecodeError:
                continue
            except FileNotFoundError:
                raise SystemExit(f"[ERROR] File not found: {self.path}")
            except Exception as exc:
                raise SystemExit(f"[ERROR] Could not read file: {exc}")
 
        if df is None or df.empty:
            raise SystemExit("[ERROR] CSV is empty or unreadable.")
 
        # Clean headers and drop fully-empty rows / columns / duplicates
        df.columns = [str(c).strip() for c in df.columns]
        df = df.dropna(how="all").dropna(axis=1, how="all")
        dupes = int(df.duplicated().sum())
        if dupes:
            self.warnings.append(f"{dupes} duplicate row(s) removed.")
            df = df.drop_duplicates()
 
        self.df = df.reset_index(drop=True)
        print(f"\n[OK] Loaded '{self.path}': "
              f"{self.df.shape[0]} rows x {self.df.shape[1]} columns")
 
    # ------------------------------------------------------------------ #
    # 2. DETECT + VALIDATE COLUMNS
    # ------------------------------------------------------------------ #
    def configure_columns(self):
        df = self.df
        obj_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()
 
        # --- name column
        named = [c for c in obj_cols if "name" in c.lower()]
        if named:
            self.name_col = named[0]
        elif obj_cols:
            self.name_col = obj_cols[0]
        else:
            self.name_col = "Student"
            df[self.name_col] = [f"Student {i + 1}" for i in range(len(df))]
        df[self.name_col] = df[self.name_col].astype(str).str.strip()
 
        # --- attendance column (coerce text like "85%" to numbers)
        for c in df.columns:
            if "attend" in c.lower():
                df[c] = pd.to_numeric(
                    df[c].astype(str).str.replace("%", "", regex=False),
                    errors="coerce")
                self.att_col = c
                break
 
        # --- subject columns: numeric (or numeric-looking), not id/total/etc.
        candidates = []
        for c in df.columns:
            if c in (self.name_col, self.att_col):
                continue
            if tokens(c) & NON_SUBJECT_TOKENS:
                continue
            converted = pd.to_numeric(df[c], errors="coerce")
            # accept the column if >= 70% of its non-null cells are numeric
            if df[c].notna().sum() and converted.notna().sum() / df[c].notna().sum() >= 0.7:
                df[c] = converted
                candidates.append(c)
        self.subject_cols = candidates
 
        if not self.subject_cols:
            raise SystemExit("[ERROR] No numeric subject columns could be identified.")
 
        # --- data quality checks
        miss = df[self.subject_cols + ([self.att_col] if self.att_col else [])].isna().sum()
        miss = miss[miss > 0]
        if not miss.empty:
            self.warnings.append(
                "Missing values (ignored in averages, not treated as zero): "
                + ", ".join(f"{k}={v}" for k, v in miss.items()))
 
        for c in self.subject_cols:
            bad = (df[c] < 0) | (df[c] > self.max_marks)
            if bad.any():
                self.warnings.append(
                    f"{int(bad.sum())} out-of-range mark(s) in '{c}' were set to missing.")
                df.loc[bad, c] = np.nan
 
        if self.att_col:
            bad = (df[self.att_col] < 0) | (df[self.att_col] > 100)
            if bad.any():
                self.warnings.append(
                    f"{int(bad.sum())} invalid attendance value(s) set to missing.")
                df.loc[bad, self.att_col] = np.nan
 
        print(f"[OK] Name column      : {self.name_col}")
        print(f"[OK] Subject columns  : {', '.join(self.subject_cols)}")
        print(f"[OK] Attendance column: {self.att_col or 'not found'}")
        for w in self.warnings:
            print(f"[WARN] {w}")
 
    # ------------------------------------------------------------------ #
    # 3. ANALYSIS
    # ------------------------------------------------------------------ #
    def analyze_data(self):
        df, subj = self.df, self.subject_cols
        marks = df[subj]
 
        # Row-wise metrics (percentage uses only subjects the student has marks for)
        df["Total_Marks"] = marks.sum(axis=1, skipna=True)
        df["Subjects_Taken"] = marks.notna().sum(axis=1)
        df["Average_Marks"] = marks.mean(axis=1)
        df["Percentage"] = (df["Total_Marks"] /
                            (df["Subjects_Taken"] * self.max_marks)) * 100
        df["Percentage"] = df["Percentage"].round(2)
 
        pct = df["Percentage"]
        df["Grade"] = pd.cut(pct, bins=GRADE_BINS, labels=GRADE_LABELS,
                             right=False).astype(object)
        df["Grade"] = df["Grade"].fillna("N/A")
        df["Result"] = np.where(pct >= self.pass_mark, "PASS", "FAIL")
 
        df["Rank"] = pct.rank(ascending=False, method="min").astype("Int64")
        df["Percentile"] = (pct.rank(pct=True) * 100).round(1)
        std = pct.std(ddof=0)
        df["Z_Score"] = ((pct - pct.mean()) / std).round(2) if std else 0.0
 
        df["Subjects_Failed"] = (marks < self.subject_pass).sum(axis=1)
        df["Best_Subject"] = marks.idxmax(axis=1)
        df["Weakest_Subject"] = marks.idxmin(axis=1)
        df["Consistency_SD"] = marks.std(axis=1).round(2)
 
        # Attendance flags + at-risk logic
        low_att = pd.Series(False, index=df.index)
        if self.att_col:
            low_att = df[self.att_col] < self.att_threshold
            df["Low_Attendance"] = low_att
        df["At_Risk"] = (pct < self.pass_mark + 10) & (
            (df["Subjects_Failed"] >= 1) | low_att | (pct < self.pass_mark))
 
        # Outliers on overall percentage (IQR rule)
        q1, q3 = pct.quantile([0.25, 0.75])
        iqr = q3 - q1
        df["Outlier"] = (pct < q1 - 1.5 * iqr) | (pct > q3 + 1.5 * iqr)
 
        # Subject statistics
        stats = pd.DataFrame({
            "Mean": marks.mean(),
            "Median": marks.median(),
            "Std": marks.std(),
            "Min": marks.min(),
            "Max": marks.max(),
            "Pass_Rate_%": (marks >= self.subject_pass).sum() / marks.notna().sum() * 100,
            "Top_Scorer": [df.loc[marks[c].idxmax(), self.name_col]
                           if marks[c].notna().any() else "-" for c in subj],
        }).round(2)
        stats.index.name = "Subject"
        self.subject_stats = stats
 
        # Subject-to-subject and attendance correlations
        corr_cols = subj + ["Percentage"] + ([self.att_col] if self.att_col else [])
        self.corr = df[corr_cols].corr()
 
        att_corr = None
        if self.att_col and df[self.att_col].notna().sum() >= 3:
            att_corr = df[self.att_col].corr(pct)
 
        top = df.loc[pct.idxmax()]
        bottom = df.loc[pct.idxmin()]
        self.insights = {
            "students": len(df),
            "mean": pct.mean(), "median": pct.median(), "std": pct.std(),
            "max": pct.max(), "min": pct.min(),
            "pass": int((df["Result"] == "PASS").sum()),
            "fail": int((df["Result"] == "FAIL").sum()),
            "top": (top[self.name_col], top["Percentage"]),
            "bottom": (bottom[self.name_col], bottom["Percentage"]),
            "best_subject": stats["Mean"].idxmax(),
            "weak_subject": stats["Mean"].idxmin(),
            "hardest_pass_rate": stats["Pass_Rate_%"].idxmin(),
            "att_corr": att_corr,
        }
 
        self._print_and_save_report()
 
    # ------------------------------------------------------------------ #
    # 4. REPORT
    # ------------------------------------------------------------------ #
    @staticmethod
    def _interpret_corr(r):
        if r is None or np.isnan(r):
            return "not enough data"
        a = abs(r)
        strength = ("very weak" if a < 0.2 else "weak" if a < 0.4 else
                    "moderate" if a < 0.6 else "strong" if a < 0.8 else "very strong")
        return f"{strength} {'positive' if r >= 0 else 'negative'}"
 
    def _print_and_save_report(self):
        df, i = self.df, self.insights
        names = lambda mask: ", ".join(df.loc[mask, self.name_col]) or "None"
 
        lines = []
        add = lines.append
        add("=" * 64)
        add("                 STUDENT PERFORMANCE REPORT")
        add("=" * 64)
        add(f"Source file            : {self.path}")
        add(f"Total students         : {i['students']}")
        add(f"Pass mark / subj. pass : {self.pass_mark:g}% / {self.subject_pass:g} marks")
        add("")
        add("-- CLASS STATISTICS " + "-" * 43)
        add(f"Mean / Median %        : {i['mean']:.2f} / {i['median']:.2f}")
        add(f"Std deviation          : {i['std']:.2f}")
        add(f"Highest / Lowest %     : {i['max']:.2f} / {i['min']:.2f}")
        add(f"Pass / Fail            : {i['pass']} / {i['fail']} "
            f"({i['pass'] / i['students'] * 100:.1f}% pass rate)")
        add(f"Top performer          : {i['top'][0]} ({i['top'][1]:.2f}%)")
        add(f"Lowest performer       : {i['bottom'][0]} ({i['bottom'][1]:.2f}%)")
        add("")
        add("-- GRADE DISTRIBUTION " + "-" * 41)
        gc = df["Grade"].value_counts().reindex(GRADE_ORDER, fill_value=0)
        for g, n in gc.items():
            add(f"  {g:<3} {'#' * n:<30} {n}")
        add("")
        add("-- SUBJECT STATISTICS " + "-" * 41)
        add(self.subject_stats.to_string())
        add("")
        add(f"Strongest subject      : {i['best_subject']}")
        add(f"Weakest subject        : {i['weak_subject']}")
        add(f"Lowest pass rate       : {i['hardest_pass_rate']}")
        add("")
        add("-- STUDENTS NEEDING ATTENTION " + "-" * 33)
        add(f"Failing overall        : {names(df['Result'] == 'FAIL')}")
        add(f"Failed >= 1 subject    : {names(df['Subjects_Failed'] >= 1)}")
        if self.att_col:
            add(f"Attendance < {self.att_threshold:g}%   : {names(df['Low_Attendance'])}")
            add(f"Attendance vs % corr.  : {i['att_corr']:.2f} "
                f"({self._interpret_corr(i['att_corr'])})"
                if i["att_corr"] is not None else "Attendance vs % corr.  : n/a")
        add(f"At-risk (combined)     : {names(df['At_Risk'])}")
        add(f"Statistical outliers   : {names(df['Outlier'])}")
        add("")
        add("-- TOP 5 RANKING " + "-" * 46)
        top5 = df.sort_values("Rank").head(5)[
            ["Rank", self.name_col, "Percentage", "Grade"]]
        add(top5.to_string(index=False))
        if self.warnings:
            add("")
            add("-- DATA QUALITY NOTES " + "-" * 41)
            lines.extend(f"  * {w}" for w in self.warnings)
        add("=" * 64)
 
        report = "\n".join(lines)
        print("\n" + report)
 
        with open("performance_report.txt", "w", encoding="utf-8") as f:
            f.write(report)
        df.to_csv("student_results.csv", index=False)
        self.subject_stats.to_csv("subject_statistics.csv")
        print("\n[OK] Saved performance_report.txt, student_results.csv "
              "and subject_statistics.csv")
 
    # ------------------------------------------------------------------ #
    # 5. VISUALS
    # ------------------------------------------------------------------ #
    def generate_visualizations(self):
        df, subj, i = self.df, self.subject_cols, self.insights
        n = len(df)
 
        fig, axs = plt.subplots(3, 3, figsize=(21, 16))
        fig.suptitle("Student Performance Dashboard", fontsize=20, fontweight="bold")
 
        # 1. Subject averages with std-dev error bars
        ax = axs[0, 0]
        s = self.subject_stats.reset_index()
        sns.barplot(data=s, x="Subject", y="Mean", hue="Subject",
                    palette="Blues_d", legend=False, ax=ax)
        for c in ax.containers:  # label bars first (error bars add a non-bar container)
            ax.bar_label(c, fmt="%.1f", padding=2, fontsize=8)
        ax.errorbar(np.arange(len(s)), s["Mean"], yerr=s["Std"].fillna(0),
                    fmt="none", ecolor="black", capsize=4, lw=1)
        ax.set_title("Subject Averages (± std dev)")
        ax.set_ylim(0, self.max_marks * 1.1)
        ax.tick_params(axis="x", rotation=30)
 
        # 2. Ranked student percentages coloured by result
        ax = axs[0, 1]
        ranked = df.sort_values("Percentage", ascending=False)
        sns.barplot(data=ranked, x=self.name_col, y="Percentage", hue="Result",
                    palette=RESULT_COLORS, dodge=False, ax=ax)
        ax.axhline(self.pass_mark, color="black", ls="--", lw=1,
                   label=f"Pass mark ({self.pass_mark:g}%)")
        ax.axhline(i["mean"], color="orange", ls=":", lw=1.5,
                   label=f"Class mean ({i['mean']:.1f}%)")
        ax.legend(fontsize=8)
        ax.set_title("Student Ranking")
        ax.set_ylim(0, 105)
        if n > 30:
            ax.set_xticks([])
        else:
            ax.tick_params(axis="x", rotation=45)
            plt.setp(ax.get_xticklabels(), ha="right")
 
        # 3. Attendance vs percentage
        ax = axs[0, 2]
        if self.att_col and df[self.att_col].notna().sum() >= 3:
            sns.regplot(data=df, x=self.att_col, y="Percentage", ax=ax,
                        color="purple", scatter_kws={"s": 70, "alpha": 0.8})
            ax.axvline(self.att_threshold, color="red", ls="--", lw=1)
            ax.text(0.03, 0.93, f"r = {i['att_corr']:.2f}", transform=ax.transAxes,
                    fontsize=11, bbox=dict(boxstyle="round", fc="white", ec="gray"))
            ax.set_title("Attendance vs Percentage")
        else:
            ax.text(0.5, 0.5, "No usable attendance data", ha="center",
                    va="center", fontsize=12)
            ax.set_title("Attendance vs Percentage")
            ax.set_axis_off()
 
        # 4. Grade distribution
        ax = axs[1, 0]
        sns.countplot(data=df, x="Grade", order=GRADE_ORDER, hue="Grade",
                      palette="viridis", legend=False, ax=ax)
        for c in ax.containers:
            ax.bar_label(c, padding=2, fontsize=9)
        ax.set_title("Grade Distribution")
        ax.set_ylabel("Students")
 
        # 5. Subject mark distributions (box + points)
        ax = axs[1, 1]
        long = df.melt(value_vars=subj, var_name="Subject", value_name="Marks")
        sns.boxplot(data=long, x="Subject", y="Marks", hue="Subject",
                    palette="Set2", legend=False, ax=ax, fliersize=0)
        sns.stripplot(data=long, x="Subject", y="Marks", color="black",
                      size=3, alpha=0.5, ax=ax)
        ax.axhline(self.subject_pass, color="red", ls="--", lw=1)
        ax.set_title("Mark Spread per Subject")
        ax.tick_params(axis="x", rotation=30)
 
        # 6. Correlation heatmap
        ax = axs[1, 2]
        sns.heatmap(self.corr, annot=True, fmt=".2f", cmap="coolwarm",
                    vmin=-1, vmax=1, square=False, ax=ax,
                    cbar_kws={"shrink": 0.8}, annot_kws={"size": 8})
        ax.set_title("Correlation Matrix")
        ax.tick_params(axis="x", rotation=30)
 
        # 7. Subject pass rate
        ax = axs[2, 0]
        pr = self.subject_stats["Pass_Rate_%"].sort_values().reset_index()
        sns.barplot(data=pr, y="Subject", x="Pass_Rate_%", hue="Subject",
                    palette="rocket_r", legend=False, ax=ax)
        for c in ax.containers:
            ax.bar_label(c, fmt="%.0f%%", padding=2, fontsize=8)
        ax.set_xlim(0, 110)
        ax.set_xlabel(f"Students scoring ≥ {self.subject_pass:g} (%)")
        ax.set_title("Subject Pass Rate")
 
        # 8. Percentage histogram + KDE
        ax = axs[2, 1]
        sns.histplot(df["Percentage"].dropna(), bins=min(10, max(3, n // 2)),
                     kde=n >= 5, color="teal", ax=ax)
        ax.axvline(i["mean"], color="orange", ls="--", label=f"Mean {i['mean']:.1f}")
        ax.axvline(i["median"], color="navy", ls=":", label=f"Median {i['median']:.1f}")
        ax.axvline(self.pass_mark, color="red", ls="-", lw=1, label="Pass mark")
        ax.legend(fontsize=8)
        ax.set_title("Percentage Distribution")
 
        # 9. Student × subject heatmap (top 25 by rank)
        ax = axs[2, 2]
        view = df.sort_values("Rank").head(25).set_index(self.name_col)[subj]
        sns.heatmap(view, annot=len(view) <= 25 and len(subj) <= 8, fmt=".0f",
                    cmap="YlGnBu", vmin=0, vmax=self.max_marks, ax=ax,
                    cbar_kws={"shrink": 0.8}, annot_kws={"size": 8})
        ax.set_ylabel("")
        ax.set_title(f"Marks Heatmap (top {len(view)} students)")
        ax.tick_params(axis="x", rotation=30)
 
        fig.tight_layout(rect=[0, 0, 1, 0.96])
        fig.savefig("dashboard.png", dpi=150, bbox_inches="tight")
        print("[OK] Dashboard saved to dashboard.png")
 
        if self.show:
            plt.show()
        plt.close(fig)
 
    # ------------------------------------------------------------------ #
    def run(self):
        self.load_data()
        self.configure_columns()
        self.analyze_data()
        self.generate_visualizations()
 
 
def ask_number(prompt, default):
    """Ask for a number; Enter keeps the default."""
    raw = input(f"{prompt} [{default:g}]: ").strip()
    try:
        return float(raw) if raw else default
    except ValueError:
        print(f"   Invalid number, using default {default:g}")
        return default
 
 
if __name__ == "__main__":
    path = input("Enter CSV file path (e.g. student_data.csv): ")
    print("\nSettings (press Enter to keep the default)")
    StudentPerformanceAnalyzer(
        path=path,
        pass_mark=ask_number("Overall pass percentage", 50.0),
        subject_pass=ask_number("Minimum marks to pass one subject", 40.0),
        max_marks=ask_number("Maximum marks per subject", 100.0),
        att_threshold=ask_number("Low-attendance threshold (%)", 75.0),
        show=True,
    ).run()
 
```
